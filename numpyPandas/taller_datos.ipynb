# Taller de NumPy y Pandas

```python
import pandas as pd
import numpy as np
```

## Ejercicio 1: Operaciones fundamentales con NumPy

```python
# Arreglo unidimensional, forma y tipo de dato
arreglo_1d = np.array([1, 2, 3, 4, 5, 6])

print("Arreglo:", arreglo_1d)
print("Shape:", arreglo_1d.shape)
print("Dtype:", arreglo_1d.dtype)
```

```python
# Cambio de forma a una matriz de 2 dimensiones
matriz_2d = arreglo_1d.reshape(2, 3)

print(matriz_2d)
print("Shape de la matriz:", matriz_2d.shape)
```

```python
# Matrices de números aleatorios
np.random.seed(42)
matriz_rand = np.random.rand(2, 3)
matriz_randint = np.random.randint(1, 10, size=(2, 3))

print("Valores entre 0 y 1 con np.random.rand():\n", matriz_rand)
print("Valores enteros con np.random.randint():\n", matriz_randint)
```

```python
# Matrices estructurales
matriz_ceros = np.zeros((3, 3))
matriz_unos = np.ones((3, 3))
matriz_identidad = np.eye(3)

print("Matriz de ceros:\n", matriz_ceros)
print("Matriz de unos:\n", matriz_unos)
print("Matriz identidad:\n", matriz_identidad)
```

```python
# Traspuesta de una matriz aleatoria
matriz_aleatoria = np.random.rand(2, 4)

print("Matriz original:\n", matriz_aleatoria)
print("Matriz traspuesta:\n", matriz_aleatoria.T)
```

```python
# Concatenación horizontal y vertical
arreglo_a = np.array([[1, 2], [3, 4]])
arreglo_b = np.array([[5, 6], [7, 8]])

print("Concatenación horizontal:\n", np.hstack((arreglo_a, arreglo_b)))
print("Concatenación vertical:\n", np.vstack((arreglo_a, arreglo_b)))
```

```python
# Operaciones matemáticas y estadísticas
datos = np.array([1, 2, 3, 4, 5])

print("Suma:", np.sum(datos))
print("Promedio:", np.mean(datos))
print("Mediana:", np.median(datos))
print("Producto:", np.prod(datos))
print("Desviación estándar:", np.std(datos))
print("Varianza:", np.var(datos))
print("Mínimo:", np.min(datos))
print("Máximo:", np.max(datos))
print("Suma acumulativa:", np.cumsum(datos))
```

```python
# Operaciones aritméticas elemento a elemento
arreglo_a = np.array([10, 20, 30])
arreglo_b = np.array([1, 2, 3])

print("Suma:", arreglo_a + arreglo_b)
print("Resta:", arreglo_a - arreglo_b)
print("Multiplicación:", arreglo_a * arreglo_b)
```

## Ejercicio 2: Integración y manipulación con Pandas

```python
# DataFrame creado desde un arreglo bidimensional de NumPy
arreglo_datos = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

df_numpy = pd.DataFrame(arreglo_datos, columns=["A", "B", "C"])
print(df_numpy)
```

```python
# DataFrame creado desde un diccionario de Python
datos_diccionario = {
    "Nombre": ["Ana", "Luis", "Marta"],
    "Edad": [25, 30, 28],
    "Ciudad": ["Bogotá", "Medellín", "Cali"]
}

df_diccionario = pd.DataFrame(datos_diccionario)
print(df_diccionario)
```

```python
# Conversión de un DataFrame a arreglos de NumPy
arreglo_con_to_numpy = df_numpy.to_numpy()
arreglo_con_values = df_numpy.values

print("Conversión con .to_numpy():\n", arreglo_con_to_numpy)
print("Conversión con .values:\n", arreglo_con_values)
```

```python
# Promedios usando np.mean() sobre el DataFrame
promedio_columnas_numpy = np.mean(df_numpy, axis=0)
promedio_filas_numpy = np.mean(df_numpy, axis=1)

print("Promedio de cada columna con np.mean():\n", promedio_columnas_numpy)
print("Promedio de cada fila con np.mean():\n", promedio_filas_numpy)
```

```python
# Promedios usando los métodos nativos de DataFrame
promedio_columnas_pandas = df_numpy.mean(axis=0)
promedio_filas_pandas = df_numpy.mean(axis=1)

print("Promedio de cada columna con df.mean(axis=0):\n", promedio_columnas_pandas)
print("Promedio de cada fila con df.mean(axis=1):\n", promedio_filas_pandas)
```
